# 1. Matrices

```python
import numpy as np
X = np.arange(1,10)
X
```

```python
matriz1 = np.reshape(X,(3,3))
matriz1
```

```python
vector1 = np.arange(1,5)
vector2 = np.arange(5,9)
vector3 = np.arange(9,13)
matriz2 = np.column_stack([vector1,vector2,vector3])
matriz2
```

```python
matriz3 = np.vstack([vector1,vector2, vector3])
matriz3
```

# Operaciones con matrices

```python
# Suma un numero a una matriz
print(matriz1 + 10)
```

```python
# multiplicacion un numero a una matriz
print(matriz1 * 10)
```

```python
# potencia a una matriz
print(matriz1 ** 3)
```

# De matrices a dataframe

```python
AUTO_1 = [20, 20, 16, 17,17,22,17]
AUTO_2 = [15,11,18,23,8,4,19]
AUTO_3 = [14,11,3,21,22,12,11]
venta_autos = np.column_stack([AUTO_1,AUTO_2, AUTO_3 ])
venta_autos
```

```python
import pandas as pd
ventas_auto_df = pd.DataFrame(venta_autos,columns=["nissan",'chevrolet','suzuki'] )
ventas_auto_df
```

```python
ventas_auto_df.index = ['1990','1991', '1992','1993', '1994','1995','1996']
ventas_auto_df
```

```python
# Acceder a un elemento especifico por posicion
ventas_auto_df.iloc[2,1]
```

```python
#Acceder a un elemento especifico mediante nombre
ventas_auto_df.loc['1990', 'nissan']
```

```python
#traer la fila 1990 y todas las columnas
ventas_auto_df.loc['1990', :]
```

```python
## años 1992 y 1993, chevrolet y suzuki
ventas_auto_df.loc[['1992','1993'],['chevrolet','suzuki']]
```

```python
## calculos de estadisticos basicos
#promedio de ventas de 1990
ventas_auto_df.loc['1990',:].mean()
## mediana de ventas de 1990
ventas_auto_df.loc['1990',:].median()
## suma de ventas de 1990
ventas_auto_df.loc['1990',:].sum()
## maximo de ventas de 1990
ventas_auto_df.loc['1990',:].max()
## minimo de ventas de 1990
ventas_auto_df.loc['1990',:].min()
```

```python
ventas_auto_df.describe()
```
